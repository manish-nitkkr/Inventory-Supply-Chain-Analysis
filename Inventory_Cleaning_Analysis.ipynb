```python
# Reload original Date column
original_dates = pd.read_csv("Inventory_SupplyChain.csv")['Date'].astype('string').str.strip()

# Create empty datetime column
parsed_dates = pd.Series(pd.NaT, index=original_dates.index, dtype='datetime64[ns]')

# 1. YYYY-MM-DD
mask = original_dates.str.match(r'^\d{4}-\d{1,2}-\d{1,2}$', na=False)
parsed_dates.loc[mask] = pd.to_datetime(
    original_dates.loc[mask],
    format='%Y-%m-%d'
)

# 2. YYYY/MM/DD
mask = original_dates.str.match(r'^\d{4}/\d{1,2}/\d{1,2}$', na=False)
parsed_dates.loc[mask] = pd.to_datetime(
    original_dates.loc[mask],
    format='%Y/%m/%d'
)

# 3. DD/MM/YYYY
mask = original_dates.str.match(r'^\d{1,2}/\d{1,2}/\d{4}$', na=False)
parsed_dates.loc[mask] = pd.to_datetime(
    original_dates.loc[mask],
    format='%d/%m/%Y'
)

# 4. MM-DD-YYYY
mask = original_dates.str.match(r'^\d{1,2}-\d{1,2}-\d{4}$', na=False)
parsed_dates.loc[mask] = pd.to_datetime(
    original_dates.loc[mask],
    format='%m-%d-%Y'
)

# Put correctly parsed dates back into df
df['Date'] = parsed_dates.loc[df.index]
```

```python
# Fill missing categorical values with mode

for col in ['Region', 'Supplier', 'Warehouse']:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Fill missing numerical values with median

numeric_cols = [
    'Inventory Level',
    'Transportation Cost',
    'Lead Time (Days)',
    'Average Inventory',
    'COGS'
]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

# Data Cleaning Summary

The Inventory & Supply Chain dataset was cleaned and prepared for further analysis and visualization.

### 1. Initial Dataset

* The dataset initially contained **10,069 rows and 17 columns**.
* The dataset includes information related to dates, regions, products, suppliers, warehouses, order status, inventory, transportation cost, lead time, backorders, warehouse capacity, and COGS.

### 2. Duplicate Records

* **52 duplicate rows** were identified initially and removed.
* After further data cleaning, **25 additional duplicate rows** were identified and removed.
* The final duplicate check confirmed **0 duplicate rows**.

### 3. Text Standardization

Text-based categorical columns were standardized using trimming and title-case formatting. The following columns were cleaned:

* Region
* Supplier
* Warehouse
* Order Status
* Category

Spelling inconsistencies such as `Noth`, `Sout`, `Suppplier A`, and `Warehoue 2` were corrected to their proper values.

### 4. Missing Values

Missing values were identified across several columns.

* Categorical missing values in **Region, Supplier, and Warehouse** were filled using the **mode**.
* Numerical missing values in **Inventory Level, Transportation Cost, Lead Time, Average Inventory, and COGS** were filled using the **median**.
* Missing values in **Order Accuracy** were filled using its **mode**.
* The final missing-value check confirmed **0 missing values across all 17 columns**.

### 5. Data Types

Data types were corrected according to the nature of each variable:

* `Date` → datetime
* `Order Accuracy` → boolean
* `Backorder` → boolean
* Transportation Cost and COGS → numeric
* Inventory, lead time, average inventory and other quantitative fields → numeric

The final dataset contains appropriate data types for analysis.

### 6. Date Cleaning

The Date column contained multiple date formats. Different formats including:

* `YYYY-MM-DD`
* `YYYY/MM/DD`
* `DD/MM/YYYY`
* `MM-DD-YYYY`

were handled and converted into a consistent datetime format.

The final date range is from **January 1, 2025 to August 31, 2026**, with no invalid or missing dates remaining.

### 7. Invalid Negative Values

Negative values were found in columns where negative values are not logically valid:

* Inventory Level
* Transportation Cost
* COGS

A total of **39 rows** contained one or more negative values. These invalid records were removed rather than replacing them with artificial values. After removal, all three columns contained **zero negative values**.

### 8. Final Dataset

After completing the cleaning process:

* **Final rows:** 9,953
* **Columns:** 17
* **Duplicate rows:** 0
* **Missing values:** 0
* **Negative invalid values:** 0

The dataset is now clean, consistent, and ready for **Exploratory Data Analysis (EDA), KPI analysis, visualization, and Power BI dashboard development**.
